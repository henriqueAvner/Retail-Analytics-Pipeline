```python
from pathlib import Path

import pandas as pd
from dotenv import dotenv_values, find_dotenv
from google.cloud import bigquery
from google.oauth2 import service_account
from pandas_gbq import to_gbq
```

```python
arquivo_env = find_dotenv(usecwd=True)
if not arquivo_env:
  raise FileNotFoundError('Crie o arquivo .env na raiz do projeto.')

caminho_credencial = dotenv_values(arquivo_env).get('GOOGLE_APPLICATION_CREDENTIALS')

if not caminho_credencial:
  raise RuntimeError(
    'Defina GOOGLE_APPLICATION_CREDENTIALS no .env com o caminho do JSON da conta de serviço.'
  )

caminho_credencial = Path(arquivo_env).parent / Path(caminho_credencial).expanduser()
credencial = service_account.Credentials.from_service_account_file(
  caminho_credencial,
  scopes=['https://www.googleapis.com/auth/bigquery']
)
```

```python
df_camp = pd.read_excel('../transform/excel_transform/campeonatos/Campeonatos.xlsx')
df_colocacoes = pd.read_excel('../transform/excel_transform/campeonatos/Colocacoes_Campeonatos.xlsx')
```

```python
df_partidas = pd.read_excel('../transform/excel_transform/partidas/Resultado_Partidas.xlsx')
```

```python
df_tier_pontuacao = pd.read_excel('../transform/excel_transform/tier/Pontuacao_tier.xlsx')
df_tier_colocacao = pd.read_excel('../transform/excel_transform/tier/Tier_Colocacao.xlsx')
```

```python
df_pt_times = pd.read_excel('../transform/excel_transform/times/Pontuacao_Times.xlsx')
df_times = pd.read_excel('../transform/excel_transform/times/Times.xlsx')
```

```python
config_gbq = {
  'projeto' : 'femeeproject',
  'dataset' : 'femeeData',
  'parametro': 'replace'
}
```

```python
print(config_gbq['parametro'])
```

```python
def import_to_gbq(df, config_gbq, credencial, table):
  to_gbq(
  dataframe=df,
  destination_table=f'{config_gbq["projeto"]}.{config_gbq['dataset']}.{table}',
  project_id=config_gbq["projeto"],
  if_exists= config_gbq["parametro"],
  credentials=credencial
)
```

```python
import_to_gbq(df_camp,config_gbq,credencial, 'Campeonatos')
import_to_gbq(df_colocacoes,config_gbq,credencial, 'Colocacoes_Campeonatos')

import_to_gbq(df_partidas,config_gbq,credencial, 'Partidas')

import_to_gbq(df_tier_pontuacao,config_gbq,credencial,'Tier_Pontuacao')
import_to_gbq(df_tier_colocacao,config_gbq,credencial,'Tier_Colocacao')

import_to_gbq(df_pt_times,config_gbq,credencial,'Pontuacao_Times')
import_to_gbq(df_times,config_gbq,credencial,'Times')
```
