```python
df_pontuacao = pd.DataFrame(pontuacao_times.get_all_values())
df_partidas = pd.DataFrame(partidas.get_all_values())
df_campeonatos = pd.DataFrame(campeonatos.get_all_values())
df_pontuacao_tier = pd.DataFrame(pontuacao_tier.get_all_values())

dfs_dict = {
  "df_pontuacao": df_pontuacao,
  "df_partidas" : df_partidas,
  "df_campeonatos" : df_campeonatos,
  "df_pontuacao_tier" : df_pontuacao_tier
}
```

```python
folder = Path("arquivos_brutos")
folder.mkdir(parents=True, exist_ok=True)



for nome, df in dfs_dict.items():

  df.to_parquet(
    path= folder / f"{nome}.parquet",
    engine='pyarrow',
    compression='snappy',
    index=None,
    partition_cols=None
  )
```
