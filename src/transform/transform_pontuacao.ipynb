```python
import pandas as pd
from pathlib import Path
from openpyxl.utils import column_index_from_string
folder = Path('../extract/arquivos_brutos/')
```

```python
df_pontuacao = pd.read_parquet(folder / 'df_pontuacao.parquet' )
df_pontuacao.head()
```

```python
df_pontuacao.columns = df_pontuacao.iloc[0]
df_pontuacao = df_pontuacao.iloc[1:].reset_index(drop=True)
```

```python
#Renomeando colunas
colunas = list(df_pontuacao.columns)
colunas[19] = "FEMEE EPT S1 RANK"
colunas[20] = "FEMEE EPT S1 PONTOS"
colunas[40] = "PTS DSF UNI-LOL"
df_pontuacao.columns = colunas

df_pontuacao.columns
```

```python
import re
import unicodedata

def normalizar_coluna(coluna):
  #Converter a coluna para string, remover espaços no inicio e no final, converter para minúsculas
  coluna = str(coluna).strip().lower()

  #Decomposição, retira acentos da palavra, ã vira a + ~
  coluna = unicodedata.normalize("NFKD", coluna)
  
  #Remove os acentos
  coluna = "".join(
    c for c in coluna
    if not unicodedata.combining(c)
  )

  coluna = re.sub(r"[^a-z0-9]+", "_", coluna)
  return coluna.strip("_")

```

```python

df_pontuacao.columns = [
  normalizar_coluna(coluna)
  for coluna in df_pontuacao.columns
]
```

```python
df_pontuacao
```

```python
df_pontuacao["ranking_atual_automatico"] = [
  float(valor.replace(",", "."))
  
  for valor in df_pontuacao["ranking_atual_automatico"]
]



print(df_pontuacao["ranking_atual_automatico"])
```

```python
df_pontuacao["penalidade_inatividade"] = [
  float(valor.replace(",", ".").replace("%", "")) / 100.0
  for valor in df_pontuacao["penalidade_inatividade"]
]
```

```python
df_pontuacao
```

```python
df_pontuacao.insert(loc=0, column='time_id', value=0)
```

```python
df_pontuacao
```

```python
df_pontuacao['nome_da_equipe'] = (
  df_pontuacao['nome_da_equipe'].str.strip()
)


df_pontuacao = df_pontuacao.loc[
  df_pontuacao['nome_da_equipe'].notna()
  & ~df_pontuacao['nome_da_equipe'].isin(['', '-'])
].copy()
```

```python
df_pontuacao['time_id'] = (
  pd.factorize(df_pontuacao['nome_da_equipe'])[0] + 1
)
```

```python
df_pontuacao['nome_da_equipe'] = [
  valor.lower().replace(' ', '_') for valor in df_pontuacao['nome_da_equipe']
]
df_pontuacao
```

```python
#Nome das colunas para a tabela de times
colunas_times = [
  "time_id",
  "nome_da_equipe",
  "ranking_atual_automatico",
  "ranking_base",
  "penalidade_inatividade"
]
```

```python
#Criando a tabela dimensao de times a partir da copia das colunas de df_pontuacao
dim_times = df_pontuacao[colunas_times].copy()
dim_times.head()
```

```python
#Renomeando colunas referentes a pontos de partida que nao seguem o padrao
df_trabalho = df_pontuacao.rename(
  columns={
    "femee_ept_s1_rank": "femee_ept_s1",
    "femee_ept_s1_pontos": "pts_femee_ept_s1"
  }
).copy()
```

```python
#Avaliando as colunas e verificando se as renomeações foram validadas
colunas_pontos = [
  coluna for coluna in df_trabalho.columns
  if coluna.startswith("pts_")
]

print(colunas_pontos)
```

```python
#Identificando colunas exclusivamente de campeonatos:
colunas_campeonatos = [
  coluna
  for coluna in df_trabalho.columns
  if coluna not in colunas_times
  and not coluna.startswith("pts_")
]
print(colunas_campeonatos)
```

```python
#Transformando os cabeçalhos de campeonatos em valores de linhas com MELT
df_colocacoes = df_trabalho.melt(
  id_vars=['time_id'],
  value_vars=colunas_campeonatos,
  var_name='campeonato',
  value_name='pontos_colocacao'
)

df_colocacoes.head(10)
```

```python
#Transformando cabeçalho de pontos de time por partida com MELT
df_pontos = df_trabalho.melt(
  id_vars=['time_id'],
  value_vars=colunas_pontos,
  var_name='campeonato',
  value_name='pontos_partidas'
)

df_pontos.head()
```

```python
#Avaliando se o melt deu certo
df_pontos.loc[df_pontos['time_id'] == 1]
```

```python
#Removendo o prefixo pts_ do melt nas colunas de campeonato, pois ja colocamos uma coluna para pontos a frente
df_pontos["campeonato"] = (
  df_pontos["campeonato"].str.removeprefix("pts_")
)
df_pontos.head(2)
```

```python
#Tratando hifen da coluna de pontos por colocacao vazios
df_colocacoes['pontos_colocacao'] = (
  df_colocacoes['pontos_colocacao'].replace("-", pd.NA)
)
df_colocacoes.head(10)
```

```python
df_pontos['pontos_partidas'] = (
  df_pontos['pontos_partidas'].replace("-", pd.NA)
)

df_pontos.head(10)
```

```python
#Transformando pontos de colocacao de string para numerico
df_colocacoes["pontos_colocacao"] = pd.to_numeric(
  df_colocacoes["pontos_colocacao"],
  errors="raise"
)
```

```python
#Transformando pontos de partida de string para numerico
df_pontos['pontos_partidas'] = pd.to_numeric(
  df_pontos['pontos_partidas'],
  errors="raise"
)
```

```python
df_colocacoes.head()
```

```python
df_pontos.head()
```

```python
#Agora, utilizando merge para juntar as 2 linhas
fato_participacoes = df_colocacoes.merge(
  df_pontos, #Junte a df_colocacoes com a df_pontos

  on=['time_id', "campeonato"], #Junte onde o time_id e o campeonato forem iguais

  how="outer", #Se nao corresponder a combinaçao, mantenha-a mesmo assim

  validate="one_to_one" #Cada correspondencia deve aparecer no max 1 vez em cada tabela
)

fato_participacoes.head(10)
```

```python
quantidade_antes = len(fato_participacoes)
quantidade_antes
```

```python
#Remove linnhas que tenham apenas TODOS os dados ausentes (how=all)
fato_participacoes = fato_participacoes.dropna(
  subset=['pontos_colocacao', 'pontos_partidas'],
  how='all'
)
```

```python
quantidade_depois = len(fato_participacoes)

print("Antes:", quantidade_antes)
print("Depois:", quantidade_depois)
print("Removidas:", quantidade_antes - quantidade_depois)
```

```python
#Resetando index apos exclusao:
fato_participacoes = fato_participacoes.reset_index(drop=True)
fato_participacoes.head(10)
```

```python
#Verificando se existem duplicatas filtrando por colunas especificas
duplicadas = fato_participacoes.duplicated(subset=['time_id', 'campeonato'], keep=False)
fato_participacoes.loc[duplicadas]
```

```python
times_cadastrados = fato_participacoes["time_id"].isin(dim_times["time_id"])
fato_participacoes.loc[~times_cadastrados]
```

```python
# Cadastra a equipe Academy caso ela ainda não exista
if not dim_times['nome_da_equipe'].eq('blackburn_ac').any():

    novo_id = int(dim_times['time_id'].max()) + 1

    nova_equipe = pd.DataFrame({
        'time_id': [novo_id],
        'nome_da_equipe': ['blackburn_ac'],
        'ranking_atual_automatico': [pd.NA],
        'ranking_base': [pd.NA],
        'penalidade_inatividade': [pd.NA]
    })

    dim_times = pd.concat(
        [dim_times, nova_equipe],
        ignore_index=True
    )
```
