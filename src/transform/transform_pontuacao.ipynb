```python
import pandas as pd
from pathlib import Path
from openpyxl.utils import column_index_from_string
folder = Path('../extract/arquivos_brutos/')
```

```python
df_pontuacao = pd.read_parquet(folder / 'df_pontuacao.parquet' )
df_pontuacao.head()
```

```python
df_pontuacao.columns = df_pontuacao.iloc[0]
df_pontuacao = df_pontuacao.iloc[1:].reset_index(drop=True)
```

```python
#Renomeando colunas
colunas = list(df_pontuacao.columns)
colunas[19] = "FEMEE EPT S1 RANK"
colunas[20] = "FEMEE EPT S1 PONTOS"
colunas[40] = "PTS DSF UNI-LOL"
df_pontuacao.columns = colunas

df_pontuacao.columns
```

```python
import re
import unicodedata

def normalizar_coluna(coluna):
  #Converter a coluna para string, remover espaços no inicio e no final, converter para minúsculas
  coluna = str(coluna).strip().lower()

  #Decomposição, retira acentos da palavra, ã vira a + ~
  coluna = unicodedata.normalize("NFKD", coluna)
  
  #Remove os acentos
  coluna = "".join(
    c for c in coluna
    if not unicodedata.combining(c)
  )

  coluna = re.sub(r"[^a-z0-9]+", "_", coluna)
  return coluna.strip("_")

```

```python

df_pontuacao.columns = [
  normalizar_coluna(coluna)
  for coluna in df_pontuacao.columns
]
```

```python
df_pontuacao
```

```python
df_pontuacao["ranking_atual_automatico"] = [
  float(valor.replace(",", "."))
  
  for valor in df_pontuacao["ranking_atual_automatico"]
]



print(df_pontuacao["ranking_atual_automatico"])
```

```python
df_pontuacao["penalidade_inatividade"] = [
  float(valor.replace(",", ".").replace("%", "")) / 100.0
  for valor in df_pontuacao["penalidade_inatividade"]
]
```

```python
df_pontuacao.insert(loc=0, column='time_id', value=0)
```

```python
df_pontuacao['nome_da_equipe'] = (
  df_pontuacao['nome_da_equipe'].str.strip()
)


df_pontuacao = df_pontuacao.loc[
  df_pontuacao['nome_da_equipe'].notna()
  & ~df_pontuacao['nome_da_equipe'].isin(['', '-'])
].copy()
```

```python
df_pontuacao['time_id'] = (
  pd.factorize(df_pontuacao['nome_da_equipe'])[0] + 1
)
```

```python
df_pontuacao['nome_da_equipe'] = [
  valor.lower().replace(' ', '_') for valor in df_pontuacao['nome_da_equipe']
]
df_pontuacao
```
