```python
import pandas as pd
from pathlib import Path

folder = Path('../extract/arquivos_brutos')

df_tier = pd.read_parquet( folder / 'df_pontuacao_tier.parquet')

df_tier
```

```python
#Tabela de regra por pontos de partida
df_pontuacao = df_tier.loc[:6]
df_pontuacao
```

```python
#Tabela de regra de pontos por colocacao
df_colocacao = df_tier.loc[7:]
df_colocacao
```

```python
df_pontuacao.columns = df_pontuacao.iloc[1]
df_pontuacao = df_pontuacao.iloc[2:].reset_index(drop=True)
```

```python
df_pontuacao
```

```python
df_colocacao
```

```python
df_colocacao.columns = df_colocacao.iloc[1]
df_colocacao = df_colocacao.iloc[2:].reset_index(drop=True)
```

```python
df_colocacao.columns.name = None
df_pontuacao.columns.name = None
```

```python
#Retirando colunas vazias dos dfs
df_pontuacao.drop(columns=[""], inplace=True)
df_colocacao.drop(columns=[""], inplace=True)
```

```python
df_pontuacao
```

```python
df_colocacao
```

```python
df_colocacao.columns = [
  valor.lower().replace(" ", "_").strip()
  for valor in df_colocacao.columns
]

df_pontuacao.columns = [
  valor.lower().replace(" ", "_").strip()
  for valor in df_pontuacao.columns
]
```

```python
df_colocacao
```

```python
df_pontuacao
```

```python
#Convertendo colunas de string para int
colunas_colocacao = df_colocacao.columns[1:]
df_colocacao[colunas_colocacao] = df_colocacao[colunas_colocacao].astype(int)

colunas_pontuacao = df_pontuacao.columns[1:]
df_pontuacao[colunas_pontuacao] = df_pontuacao[colunas_pontuacao].astype(int)


```

```python
#Normalizando a coluna TIER para conter apenas a letra do tier
df_colocacao['tier'] = [
  valor.upper().replace("tier ", "")
  for valor in df_colocacao['tier']
]


df_pontuacao['tier'] = [
  valor.upper().replace("TIER ", "")
  for valor in df_pontuacao['tier']
]
```

```python
df_colocacao
```

```python
#Fazendo unpivot da tabela de colocacao

df_colocacao_long = df_colocacao.melt(
  id_vars="tier",       #Definindo o que será o identificador da tabela
  var_name="colocacao", #coluna que vai pegar o nome de todas as outras colunas, e transformar em linhas
  value_name="pontos"   #Nome da coluna que vai receber os valores que preenchem as colunas antigas
)

```

```python
df_colocacao_long.head()
```

```python
df_pontuacao_long = df_pontuacao.melt(
  id_vars="tier",
  var_name="formato",
  value_name= "pontos"
)
```

```python

df_pontuacao_long.head()
```

```python
rename_columns = {
  '1º' : '1º lugar',	
  '2º' : '2º lugar',		
  '3º' : '3º lugar',		
  '4º' : '4º lugar',		
  '5º' : '5º lugar',		
  '6º' : '6º lugar',		
  '7º' : '7º lugar',		
  '8º' : '8º lugar'
}

df_colocacao_long['colocacao'] = df_colocacao_long['colocacao'].replace(rename_columns)
```
